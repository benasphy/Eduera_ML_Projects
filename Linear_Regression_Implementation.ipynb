**Predicting House Prices Using Linear Regression in Python**

Welcome! In this tutorial, we’ll learn how to use Linear Regression to predict house prices based on their size. We'll be using Python with popular libraries like NumPy, Scikit-Learn, and Matplotlib. This is a beginner-friendly example that introduces how machine learning models are trained and used for prediction.

We will:

1. Prepare the data.
2. Fit a linear regression model.
3. Make predictions.
4. Visualize the results.

Step 1: Import Necessary Libraries

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import pandas as pd
```

Step 2: Define the Dataset

```python
# Example data: House sizes and prices
house_sizes = np.array([1400, 1600, 1700, 1875, 1100, 1550, 2350, 2450, 1425, 1700])
house_prices = np.array([245, 312, 279, 308, 199, 219, 405, 324, 319, 255])

df = pd.DataFrame({
    'House Size (sq.ft)' : house_sizes,
    'House Price ($1000)' : house_prices
})
df
```

Step 3: Reshape the Data

```python
#reshape the data to 2D arrays

X = house_sizes.reshape(-1,1)
y = house_prices
```

Step 4: Create and Train the Model

```python
#create a Linear Regression model
model = LinearRegression()

#Fit the model to the data
model.fit(X, y)
```

Step 5: Make Predictions

```python
#Predict house prices for new sizes

new_sizes = np.array([2000, 1800, 2100]).reshape(-1, 1)
predictions = model.predict(new_sizes)
predictions #print out the predicted prices
```

Step 6: Visualize the Data and Predictions

```python
#plot the data and Regression line

plt.scatter(house_sizes, house_prices, color = 'blue', label = "Actual Data")
plt.plot(house_sizes, model.predict(X), color = 'red', label = "Regression Line")
plt.scatter(new_sizes, predictions, color = "green", label = "Predictions")
plt.xlabel('House Size (sq.ft.)')
plt.ylabel('House Price ($1000)')
plt.legend()
plt.show()
```

Step 7: Print the Predictions

```python
#Display Predictions

for size, prediction in zip(new_sizes.flatten(), predictions):
  print(f"Predicted Price for a {size} sq.ft. house in($1000): ${prediction:.2f}")
```
