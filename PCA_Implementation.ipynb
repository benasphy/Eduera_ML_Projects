```python
#Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
```

```python
#Load the iris dataset (150 samples, 4 features that means 4D shape)
iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names
print(X.shape)
```

```python
#Create PCA model to reduce from 4D to 2D
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)
print(X_reduced.shape)
```

```python
# Explained Variance (how much info is kept)
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total variance explained by 2 components:", pca.explained_variance_ratio_.sum())
```

```python
#Plot the reduced 2D data
plt.figure(figsize=(6,5))
colors = ["red", "green", "blue"]

for color, i, target_name in zip(colors, [0,1,2], target_names):
  plt.scatter(
      X_reduced[y == i, 0], X_reduced[y == i, 1],
      color = color, alpha=0.6, label=target_name
  )
plt.legend()
plt.title("Iris Dataset reduced from 4D to 2D")
plt.xlabel("PCA 1")
plt.ylabel("PCA2")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
